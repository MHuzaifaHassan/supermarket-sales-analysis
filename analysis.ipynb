```python
branch_sales = df.groupby("Branch")["Total"].sum()

plt.figure(figsize=(8, 5))
branch_sales.plot(kind="bar", color=["blue", "green", "red"])
plt.title("Branch wise Total Sales")
plt.xlabel("Branch")
plt.ylabel("Total Sales")
plt.xticks(rotation=0)
plt.show()
```

```python
product_sales = df.groupby("Product line")["Total"].sum().reset_index()

plt.figure(figsize=(10, 5))
sns.barplot(data=product_sales, x="Product line", y="Total", palette="coolwarm")
plt.title("Product Line wise Sales")
plt.xlabel("Product Line")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.show()
```

```python
gender_sales = df.groupby("Gender")["Total"].sum()

plt.figure(figsize=(6, 6))
plt.pie(gender_sales, labels=gender_sales.index, autopct="%1.1f%%", colors=["pink", "blue"])
plt.title("Sales by Gender")
plt.show()
```

```python
monthly_sales = df.groupby("Month")["Total"].sum()

plt.figure(figsize=(8, 5))
plt.plot(monthly_sales.index, monthly_sales.values, marker="o", color="blue", linewidth=2)
plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.xticks([1, 2, 3], ["January", "February", "March"])
plt.show()
```

```python
ratings = np.array(df["Rating"])

plt.figure(figsize=(8, 5))
plt.hist(ratings, bins=10, color="purple", edgecolor="black")
plt.title("Customer Rating Distribution")
plt.xlabel("Rating")
plt.ylabel("Number of Customers")
plt.axvline(np.mean(ratings), color="red", linestyle="dashed", linewidth=2, label=f"Mean: {np.mean(ratings):.2f}")
plt.legend()
plt.show()
```

```python
correlation = df[["Total", "Quantity", "Rating", "Unit price", "gross income"]].corr()

plt.figure(figsize=(8, 6))
sns.heatmap(correlation, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Heatmap")
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x="Rating", y="Total", hue="Branch", palette="Set1")
plt.title("Rating vs Total Sales")
plt.xlabel("Rating")
plt.ylabel("Total Sales")
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.boxplot(data=df, x="Branch", y="Rating", palette="Set2")
plt.title("Branch wise Rating Distribution")
plt.xlabel("Branch")
plt.ylabel("Rating")
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.countplot(data=df, x="Payment", palette="Set3")
plt.title("Payment Method Distribution")
plt.xlabel("Payment Method")
plt.ylabel("Number of Customers")
plt.show()
```
